## Load the spaCy Model and the Article

Load the pretrained English model into memory, then read the article text from `article.txt`.

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("Model loaded successfully.")

with open("article.txt", "r", encoding="utf-8") as f:
    text = f.read()

print(f"Loaded article with {len(text.split())} words.\n")
print(text[:300], "...")
```

## Perform POS Tagging

Running the text through the spaCy pipeline tokenizes it and assigns a POS tag to every single token.

```python
doc = nlp(text)

print("Sample of POS-tagged tokens:\n")
for token in list(doc)[:20]:
    print(f"{token.text:<15} {token.pos_}")
```

## Extract Nouns and Verbs

Only keep tokens tagged NOUN or VERB. Token needs to be alphabetic, which filters out punctuation, numbers, and whitespace automatically.

```python
target_pos = {"NOUN", "VERB"}

extracted_words = [
    (token.text, token.pos_)
    for token in doc
    if token.pos_ in target_pos and token.is_alpha
]

print(f"Total nouns and verbs extracted: {len(extracted_words)}")
print("\nFirst 15 extracted words:")
for word, pos in extracted_words[:15]:
    print(f"{word:<15} {pos}")
```

## Save Extracted Words to CSV

```python
import csv

OUTPUT_CSV = "YomineshGiri_POS_01.csv"

with open(OUTPUT_CSV, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Word", "POS_Tag"])
    writer.writerows(extracted_words)

print(f"Saved {len(extracted_words)} rows to {OUTPUT_CSV}")
```

## Conclusion

For this assignment, a BBC News article titled "How Wimbledon cracked cricket-obsessed India" was used and processed through spaCy's POS tagger. After loading the article as a text file, the pretrained `en_core_web_sm` model was used to tag every word in the article with its part of speech.

Once every word had a tag the results were filtered to keep only nouns and verbs. Out of 863 words in the article, 331 were extracted as nouns or verbs and saved into a CSV file with two columns: Word and POS_Tag.

Overall, this task demonstrated how POS tagging works in practice i.e. the model doesn't just look words up in a list, it predicts each tag based on the context of the sentence which is why the same word can be tagged differently depending on how it's used.
